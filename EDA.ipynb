Key Characteristics & Metrics

Focus: Analyzes single-feature properties like central tendency (mean, median, mode), dispersion (variance, standard deviation, range), and frequency distribution.

BOX PLOT

A box plot is a standardized graphical method for displaying the distribution of numerical data based on its five-number summary: minimum, first quartile, median, third quartile, and maximum. These are plotted to visualize the distribution, spread of data and to identify outliers in data.

```python
#Box plot of imdb_score
plt.figure(figsize=(8,6))
sns.boxplot(data=data, y='imdb_score')
plt.title('Box plot of imdb_score\n', color='brown')
plt.show()
```

1. Why a box plot?
A box plot was chosen to examine the distribution and spread of imdb_score, and to specifically identify statistical outliers (exceptionally high or low-rated titles) using the five-number summary (median, quartiles, and whiskers).

2. What are the insight(s) found from the chart?

The median IMDb score sits around 6.1, with the central 50% of titles (the interquartile range) generally falling between roughly 5.2 and 6.8.

The box plot reveals a significant number of outlier data points extending below the lower whisker (poorly rated titles) and above the upper whisker (exceptionally acclaimed masterpieces).

3. Will the gained insights help creating a positive business impact? Are there any insights that lead to negative growth?
Yes, identifying rating distributions and outliers allows content curators to filter out low-quality titles that risk degrading user satisfaction while spotlighting top-tier content for promotional campaigns. Recognizing these quality tiers prevents investing in underperforming genres and directly optimizes content recommendation algorithms for higher user engagement.

BIVARIATE ANALYSIS
Bivariate analysis is a statistical method used to examine the relationship, correlation, or association between two different variables simultaneously to determine how they influence or relate to one another.

Key Characteristics & Metrics

Focus: Analyzes pairwise interactions, dependencies, or associations between two distinct features (e.g., numerical vs. numerical, categorical vs. categorical, or numerical vs. categorical).

Common Visualizations: Uses scatter plots, grouped bar charts, box plots across categories, correlation heatmaps, and pair plots.

Primary Objective: Identifies patterns, trends, causality, or multicollinearity between features to build predictive models and extract actionable business intelligence.

SCATTER PLOT
A scatter plot is a bivariate visualization tool that displays individual data points on a two-dimensional Cartesian plane using X and Y axes to examine the relationship or correlation between two numerical variables.

Key Characteristics & Use Cases

Relationship Assessment: Helps quickly reveal whether two variables have a positive correlation, negative correlation, or no correlation at all.

Pattern & Outlier Detection: Identifies clusters, trends, nonlinear patterns, and anomalous data points or outliers between two features.

Regression Analysis: Often paired with a line of best fit or trendline to visually evaluate the strength and direction of the association for predictive modeling.

VIOLIN PLOT
A violin plot is a statistical data visualization that combines a box plot with a kernel density plot to display the distribution, probability density, median, and interquartile range of numerical data across different categories.

Key Characteristics & Use Cases

Distribution Shape: Unlike standard box plots that only show summary statistics (median and quartiles), violin plots reveal the full underlying distribution shape, including multi-modality or skewness.

Comparative Analysis: Highly effective for comparing multiple categories or groups side-by-side to see how numerical values are spread across different segments.

Symmetrical Profile: The width of the shaded area (the "violin") at any given point represents the proportion of data points located at that specific value.

Diagonal Distributions: The diagonal plots typically display histograms or kernel density estimates (KDE) to show the frequency distribution of each specific feature.

Exploratory Data Analysis (EDA): Widely used in machine learning workflows to quickly identify linear or non-linear correlations, group clusters, and spot outliers across multiple dimensions prior to modeling.

Correlation Matrices: Frequently applied in multivariable analysis to display correlation coefficients between multiple numerical variables, making it easy to spot strong positive or negative associations.

Pattern & Hotspot Recognition: Enables analysts to visually isolate clusters, seasonal trends, anomalies, or high-density concentration zones within complex datasets at a glance.
