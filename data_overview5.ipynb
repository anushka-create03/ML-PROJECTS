```python
import pandas as pd
import matplotlib.pyplot as plt

# Recalculate price per square foot in rupees
data["Calculated_Price_per_SqFt"] = (
    data["Price_in_Lakhs"] * 100000 / data["Size_in_SqFt"]
)

# Amenities to compare
amenities_to_check = ["Gym", "Pool", "Garden", "Clubhouse", "Playground"]

# Split the combined Amenities text into individual amenity names
amenity_lists = data["Amenities"].fillna("").apply(
    lambda value: [item.strip() for item in str(value).split(",")]
)

# Calculate average price per sq ft for properties with and without each amenity
results = []

for amenity in amenities_to_check:
    has_amenity = amenity_lists.apply(lambda items: amenity in items)

    results.append({
        "Amenity": amenity,
        "With amenity": data.loc[
            has_amenity, "Calculated_Price_per_SqFt"
        ].mean(),
        "Without amenity": data.loc[
            ~has_amenity, "Calculated_Price_per_SqFt"
        ].mean(),
    })

chart_data = pd.DataFrame(results).set_index("Amenity")

# Draw the grouped bar chart
ax = chart_data.plot(
    kind="bar",
    figsize=(10, 6),
    color=["seagreen", "lightcoral"]
)

ax.set_title("Average Price per Sq Ft With and Without Amenities")
ax.set_xlabel("Amenity")
ax.set_ylabel("Average price per sq ft (₹)")
ax.legend(title="Property group")
plt.xticks(rotation=0)
plt.tight_layout()

# Save the image for your report
plt.savefig("avg_price_per_sqft_by_amenity.png", dpi=300, bbox_inches="tight")
plt.show()

# Display the averages used in the chart
display(chart_data.round(2))
```

```python
annual_growth_rate = 0.08  # assuming 8% per year
years = 5

clean_data["Estimated_Price_After_5_Years"] = (
    clean_data["Price_in_Lakhs"] * (1 + annual_growth_rate) ** years
)

display(
    clean_data[
        ["Price_in_Lakhs", "Estimated_Price_After_5_Years"]
    ].head()
)
```

```python
# Calculate using full precision instead of the rounded column
calculated_price_per_sqft = (
    clean_data["Price_in_Lakhs"] / clean_data["Size_in_SqFt"]
)

median_price_per_sqft = calculated_price_per_sqft.median()

clean_data["Good_Investment"] = (
    calculated_price_per_sqft <= median_price_per_sqft
).astype(int)

print("Median price per sq ft:", round(median_price_per_sqft, 4))
print("\nInvestment label counts (0 = No, 1 = Yes):")
print(clean_data["Good_Investment"].value_counts().sort_index())
```

```python
columns_to_leave_out = [
    "ID",
    "Price_in_Lakhs",
    "Price_per_SqFt",
    "Estimated_Price_After_5_Years",
    "Good_Investment",
    "Floor_No",
    "Total_Floors",
    "Floor_Data_Review",
]

X = clean_data.drop(columns=columns_to_leave_out)
y = clean_data["Good_Investment"]

print("Input data shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget counts (0 = No, 1 = Yes):")
print(y.value_counts().sort_index())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Checking rows:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

categorical_columns = X_train.select_dtypes(
    include=["object", "string"]
).columns.tolist()

numeric_columns = X_train.select_dtypes(
    include="number"
).columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ("numbers", StandardScaler(), numeric_columns),
        ("categories", OneHotEncoder(handle_unknown="ignore"), categorical_columns),
    ]
)

model = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        ("classifier", LogisticRegression(max_iter=1000)),
    ]
)

print("Categorical columns:", categorical_columns)
print("Numeric columns:", numeric_columns)
```

```python
model.fit(X_train, y_train)
print("Model training finished.")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

predictions = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, predictions), 3))
print("\nClassification report:")
print(classification_report(y_test, predictions))
print("\nConfusion matrix:")
print(confusion_matrix(y_test, predictions))
```

```python
baseline_accuracy = y_test.value_counts(normalize=True).max()
model_accuracy = accuracy_score(y_test, predictions)

print("Baseline accuracy:", round(baseline_accuracy, 3))
print("Model accuracy:", round(model_accuracy, 3))
```

```python
from sklearn.model_selection import train_test_split

regression_target = "Estimated_Price_After_5_Years"

columns_to_leave_out = [
    "ID",
    "Price_per_SqFt",
    "Floor_No",
    "Total_Floors",
    "Floor_Data_Review",
    "Good_Investment",
    regression_target,
]

X_reg = clean_data.drop(columns=columns_to_leave_out)
y_reg = clean_data[regression_target]

X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42
)

print("Training rows:", len(X_reg_train))
print("Checking rows:", len(X_reg_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

categorical_columns_reg = X_reg_train.select_dtypes(
    include=["object", "string"]
).columns.tolist()

numeric_columns_reg = X_reg_train.select_dtypes(
    include="number"
).columns.tolist()

regression_preprocessor = ColumnTransformer(
    transformers=[
        ("numbers", StandardScaler(), numeric_columns_reg),
        ("categories", OneHotEncoder(handle_unknown="ignore"), categorical_columns_reg),
    ]
)

regression_model = Pipeline(
    steps=[
        ("preprocessing", regression_preprocessor),
        ("regressor", LinearRegression()),
    ]
)

regression_model.fit(X_reg_train, y_reg_train)
print("Price model training finished.")
```

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

price_predictions = regression_model.predict(X_reg_test)

mae = mean_absolute_error(y_reg_test, price_predictions)
rmse = np.sqrt(mean_squared_error(y_reg_test, price_predictions))
r2 = r2_score(y_reg_test, price_predictions)

print("MAE (lakhs):", round(mae, 2))
print("RMSE (lakhs):", round(rmse, 2))
print("R² score:", round(r2, 3))
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import joblib

investment_model = joblib.load("investment_model.pkl")
price_model = joblib.load("price_model.pkl")

st.title("Real Estate Investment Advisor")
st.write("Enter property details to see the model estimates.")

state = st.text_input("State", "Tamil Nadu")
city = st.text_input("City", "Chennai")
locality = st.text_input("Locality", "Locality_84")
property_type = st.selectbox(
    "Property type", ["Apartment", "Villa", "Independent House"]
)
bhk = st.number_input("Bedrooms (BHK)", min_value=1, max_value=10, value=2)
size = st.number_input("Size (sq ft)", min_value=100, max_value=20000, value=1000)
price = st.number_input("Current price (lakhs)", min_value=1.0, value=100.0)
year_built = st.number_input(
    "Year built", min_value=1900, max_value=2025, value=2010
)
furnished = st.selectbox(
    "Furnished status", ["Unfurnished", "Semi-furnished", "Furnished"]
)
schools = st.number_input("Nearby schools (dataset value)", min_value=0, value=5)
hospitals = st.number_input("Nearby hospitals (dataset value)", min_value=0, value=5)
transport = st.selectbox(
    "Public transport access", ["Low", "Medium", "High"]
)
parking = st.selectbox("Parking", ["No", "Yes"])
security = st.selectbox("Security", ["No", "Yes"])
amenities = st.text_input("Amenities", "Gym, Garden")
facing = st.selectbox("Facing", ["North", "South", "East", "West"])
owner = st.selectbox("Owner type", ["Owner", "Builder", "Broker"])
availability = st.selectbox(
    "Availability", ["Ready_to_Move", "Under_Construction", "Sold"]
)

if st.button("Get estimate"):
    age = 2025 - year_built

    property_data = {
        "State": state,
        "City": city,
        "Locality": locality,
        "Property_Type": property_type,
        "BHK": bhk,
        "Size_in_SqFt": size,
        "Year_Built": year_built,
        "Furnished_Status": furnished,
        "Age_of_Property": age,
        "Nearby_Schools": schools,
        "Nearby_Hospitals": hospitals,
        "Public_Transport_Accessibility": transport,
        "Parking_Space": parking,
        "Security": security,
        "Amenities": amenities,
        "Facing": facing,
        "Owner_Type": owner,
        "Availability_Status": availability,
    }

    classification_input = pd.DataFrame([property_data])
    investment_result = investment_model.predict(classification_input)[0]

    regression_data = property_data.copy()
    regression_data["Price_in_Lakhs"] = price
    regression_input = pd.DataFrame([regression_data])
    future_price = price_model.predict(regression_input)[0]

    st.subheader("Estimates")
    st.write(
        "Good investment"
        if investment_result == 1
        else "Not marked as a good investment"
    )
    st.write(f"Estimated price after 5 years: ₹{future_price:.2f} lakhs")
    st.caption(
        "These are project estimates based on the dataset's rules, "
        "including an assumed 8% annual price increase."
    )
```

```python
from xgboost import XGBClassifier
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# XGBoost doesn't need numeric scaling, so pass numbers through unchanged
xgb_preprocessor = ColumnTransformer(
    transformers=[
        ("numbers", "passthrough", numeric_columns),
        ("categories", OneHotEncoder(handle_unknown="ignore"), categorical_columns),
    ]
)

xgb_model = Pipeline(
    steps=[
        ("preprocessing", xgb_preprocessor),
        ("classifier", XGBClassifier(
            n_estimators=200,
            max_depth=6,
            learning_rate=0.1,
            subsample=0.8,
            colsample_bytree=0.8,
            tree_method="hist",
            eval_metric="logloss",
            random_state=42,
            n_jobs=-1,
        )),
    ]
)

xgb_model.fit(X_train, y_train)
xgb_predictions = xgb_model.predict(X_test)

print("XGBoost accuracy:", round(accuracy_score(y_test, xgb_predictions), 3))
print("\nClassification report:")
print(classification_report(y_test, xgb_predictions))
print("\nConfusion matrix:")
print(confusion_matrix(y_test, xgb_predictions))
```
